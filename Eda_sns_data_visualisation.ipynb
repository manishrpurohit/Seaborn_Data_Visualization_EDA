# Comprehensive Exploratory Data Analysis (EDA)

As a Senior Analyst and Data Scientist, I present this Exploratory Data Analysis covering 65 diverse chart types. Over 30 of these are fully implemented in Python using standard datasets (`diamonds`, `tips`, and `flights`) from Seaborn.


## Descriptions of 65 Charts/Diagrams
1. **Histogram plot**: Shows frequency distribution of continuous data. Insight: Skewness, central tendency.
2. **Density plot**: Smoothed version of a histogram. Insight: Underlying continuous probability distribution.
3. **KDE plot**: Kernel Density Estimate, estimates the probability density function. Insight: Smooth distribution shape.
4. **Pie chart**: Circular chart for categorical proportions. Insight: Relative sizes of parts to a whole.
5. **Donut chart**: Pie chart with a hole. Insight: Proportional data with space for a central label.
6. **Line chart**: Data points connected by lines. Insight: Trends over time or ordered sequences.
7. **Multi-line chart**: Multiple lines on one chart. Insight: Comparing trends between different groups.
8. **Bar plot**: Rectangular bars for categorical data. Insight: Comparing magnitudes across categories.
9. **Dual-axis line chart**: Line chart with two Y-axes. Insight: Comparing two metrics with different scales over time.
10. **Stacked bar plot**: Bars stacked on each other. Insight: Part-to-whole relationships across categories.
11. **Clustered barplot**: Grouped bars side-by-side. Insight: Comparing sub-categories within main categories.
12. **Heat map**: Matrix of colors representing values. Insight: Correlations, patterns in 2D cross-tabulations.
13. **Tree map**: Nested rectangles. Insight: Hierarchical data proportions and clustering.
14. **Correlation map**: Heatmap of a correlation matrix. Insight: Linear relationships between numerical variables.
15. **Sunburst diagram**: Multilevel pie chart. Insight: Hierarchical proportional data radiating outwards.
16. **Scatter plot**: Points plotted on X and Y axes. Insight: Relationship or correlation between two numerical variables.
17. **Dot plot**: Points on a single axis. Insight: Distribution of small datasets, identifying gaps/clusters.
18. **Bubble chart**: Scatter plot with variable point sizes. Insight: Relationships across three numerical variables.
19. **Lollipop chart**: Bar chart with lines and dots. Insight: Clean visual comparison of categorical magnitudes.
20. **Radar chart**: Spider web plot with multiple axes. Insight: Comparing multiple variables for a single subject.
21. **Violin plot**: Combines box plot and KDE. Insight: Distribution, median, and density across categories.
22. **Box plot**: Displays quartiles and outliers. Insight: Spread, skewness, and extreme values.
23. **Whisker plot**: The 'whisker' part of a box plot. Insight: Range of the bulk of the data (1.5 IQR).
24. **Waterfall chart**: Cumulative effect of sequential values. Insight: Understanding how a value changed incrementally.
25. **Ridge plot**: Overlapping density plots. Insight: Comparing distributions across many categories simultaneously.
26. **Network chart**: Nodes and edges. Insight: Relationships and structures in graphs or webs.
27. **Word cloud**: Text visualization where size indicates frequency. Insight: Identifying most common terms in text.
28. **Swarm plot**: Categorical scatter plot without overlap. Insight: Granular distribution of data points.
29. **Venn diagram**: Overlapping circles. Insight: Logical sets and intersections between groups.
30. **Dendrogram**: Tree-like diagram. Insight: Hierarchical clustering relationships and distances.
31. **Area chart**: Filled line chart. Insight: Volume or magnitude accumulation over time.
32. **Stacked area chart**: Multiple filled areas. Insight: Part-to-whole composition changes over time.
33. **Chord diagram**: Connections between entities in a circle. Insight: Flows or bilateral relationships.
34. **Sankey chart**: Flow diagram with varying widths. Insight: Flow quantities and transitions between stages.
35. **Arc diagram**: Nodes on a line with arcs connecting them. Insight: Network connections and sequences.
36. **Gantt chart**: Horizontal bar chart for schedules. Insight: Project timelines and task dependencies.
37. **Parallel plot**: Lines crossing parallel axes. Insight: Multivariate data patterns and clustering.
38. **Choropleth map**: Map with shaded regions. Insight: Geographic data distributions and densities.
39. **Bump chart**: Line chart for rankings. Insight: Changes in rank or position over time.
40. **Connection map**: Map with lines between points. Insight: Geographic flows or origin-destination routes.
41. **Hexbin map**: Hexagonal binning scatter plot. Insight: Density of points in continuous 2D space.
42. **Stream graph**: Centered stacked area chart. Insight: Volume changes over time with smooth organic visualization.
43. **Circular bar chart**: Bar chart in a polar coordinate system. Insight: Cyclic data or aesthetically pleasing comparisons.
44. **Polar area chart**: Pie chart with varying radius. Insight: Cyclical patterns and magnitudes.
45. **Multilevel pie/donut chart**: Concentric rings. Insight: Deep hierarchical proportions.
46. **Hive plot**: Network diagram on axes. Insight: Structured network analysis minimizing edge crossing.
47. **Flow plot**: General term for Sankey/network flows. Insight: Movement of quantities through systems.
48. **Pictorial chart**: Using icons to represent data. Insight: Engaging proportional visual for non-technical audiences.
49. **Combo chart**: Combines line and bar charts. Insight: Mixed metrics comparison (e.g., volume and rate).
50. **Pareto chart**: Bar and cumulative line chart. Insight: Identifying largest factors (80/20 rule).
51. **Pyramid chart**: Triangle or bi-directional chart. Insight: Hierarchical structure or demographic populations.
52. **Euler's chart**: Like Venn but area-proportional. Insight: Realistic set overlaps where missing intersections are omitted.
53. **Slope chart**: Line chart between two discrete points. Insight: Before/after changes or rank shifts.
54. **Process diagram**: Flowchart. Insight: Logical steps in a process or decision tree.
55. **Column chart**: Vertical bar chart. Insight: Comparing categorical values vertically.
56. **Stacked column chart**: Vertical stacked bars. Insight: Part-to-whole in categories vertically.
57. **Grouped column chart**: Vertical clustered bars. Insight: Comparing sub-categories within groups vertically.
58. **Waffle chart**: Grid of squares. Insight: Percentage progress or exact discrete proportions.
59. **Q-Q plot**: Quantile-Quantile plot. Insight: Checking if data follows a theoretical distribution (e.g., normal).
60. **Control chart**: Line chart with control limits. Insight: Process stability over time in manufacturing/quality.
61. **Strip plot**: 1D scatter plot. Insight: Distribution of individual points along an axis.
62. **Jitter plot**: Strip plot with random noise. Insight: Reducing overlap in strip plots to see density.
63. **Curve bar chart**: Bar chart with rounded corners. Insight: Modern aesthetic variant of the bar chart.
64. **Drilldown/Drillup chart**: Interactive hierarchical chart. Insight: Zooming into nested data levels dynamically.
65. **Timer**: Gauge or progress visualization. Insight: Time elapsed or temporal progress against a goal.


## Setup and Data Loading
Loading standard datasets for the implementations. `seaborn` provides robust mock datasets we will use.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import warnings
warnings.filterwarnings('ignore')

# Load standard datasets
diamonds = sns.load_dataset('diamonds')
tips = sns.load_dataset('tips')
flights = sns.load_dataset('flights')
np.random.seed(42)
```

### 1. Histogram Plot
Insight: Identifies the distribution of diamond carats.


```python
plt.figure(figsize=(8, 5))
sns.histplot(diamonds['carat'], bins=30, kde=False, color='skyblue')
plt.title('1. Histogram of Diamond Carats')
plt.xlabel('Carat')
plt.ylabel('Frequency')
plt.show()
```

### 2. Density Plot
Insight: Shows the smooth probability density of diamond depth.


```python
plt.figure(figsize=(8, 5))
sns.kdeplot(diamonds['depth'], fill=True, color='orange')
plt.title('2. Density Plot of Diamond Depth')
plt.xlabel('Depth')
plt.ylabel('Density')
plt.show()
```

### 3. KDE Plot
Insight: Multiple KDE distributions for comparison based on sex.


```python
plt.figure(figsize=(8, 5))
sns.kdeplot(data=tips, x='total_bill', hue='sex', fill=True, common_norm=False)
plt.title('3. KDE Plot of Total Bill by Gender')
plt.show()
```

### 4. Pie Chart
Insight: Proportional view of diamond cut quality categories.


```python
plt.figure(figsize=(6, 6))
cut_counts = diamonds['cut'].value_counts()
plt.pie(cut_counts, labels=cut_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('4. Pie Chart: Diamond Cuts')
plt.show()
```

### 5. Donut Chart
Insight: A pie chart with a hole, good for emphasizing proportions cleanly.


```python
plt.figure(figsize=(6, 6))
day_counts = tips['day'].value_counts()
plt.pie(day_counts, labels=day_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('Set2'))
centre_circle = plt.Circle((0,0),0.70,fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)
plt.title('5. Donut Chart: Days of Week')
plt.show()
```

### 6. Line Chart
Insight: Aggregated passengers over time showing general upward trend.


```python
plt.figure(figsize=(10, 5))
flights_sum = flights.groupby('year')['passengers'].sum().reset_index()
sns.lineplot(data=flights_sum, x='year', y='passengers', marker='o')
plt.title('6. Line Chart: Total Passengers by Year')
plt.grid(True)
plt.show()
```

### 7. Multi-line Chart
Insight: Compares time series across multiple categories (months).


```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=flights, x='year', y='passengers', hue='month')
plt.title('7. Multi-line Chart: Passengers by Month over Years')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

### 8. Bar Plot
Insight: Categorical average total bill per day.


```python
plt.figure(figsize=(8, 5))
sns.barplot(data=tips, x='day', y='total_bill', errorbar=None, palette='viridis')
plt.title('8. Bar Plot: Average Total Bill by Day')
plt.show()
```

### 9. Dual-axis Line Chart
Insight: Shows two differing metrics (Price vs Carat) on dual y-axes.


```python
cut_stats = diamonds.groupby('cut', observed=False)[['price', 'carat']].mean().reset_index()
fig, ax1 = plt.subplots(figsize=(10, 5))
ax2 = ax1.twinx()
ax1.plot(cut_stats['cut'], cut_stats['price'], 'g-o', label='Average Price')
ax2.plot(cut_stats['cut'], cut_stats['carat'], 'b-s', label='Average Carat')
ax1.set_xlabel('Cut')
ax1.set_ylabel('Average Price ($)', color='g')
ax2.set_ylabel('Average Carat', color='b')
plt.title('9. Dual-axis Line Chart')
plt.show()
```

### 10. Stacked Bar Plot
Insight: Parts of a whole (smokers vs non) over different days.


```python
day_smoker = pd.crosstab(tips['day'], tips['smoker'])
day_smoker.plot(kind='bar', stacked=True, figsize=(8, 5), colormap='Set1')
plt.title('10. Stacked Bar Plot: Smokers per Day')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

### 11. Clustered Barplot (Grouped Column)
Insight: Comparisons within groups (Sex) within categories (Day).


```python
plt.figure(figsize=(8, 5))
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', errorbar=None, palette='muted')
plt.title('11. Clustered Barplot: Total Bill by Day and Sex')
plt.show()
```

### 12. Heat Map
Insight: 2D representation showing hot/cold spots for passenger volume.


```python
flights_pivot = flights.pivot(index='month', columns='year', values='passengers')
plt.figure(figsize=(10, 8))
sns.heatmap(flights_pivot, annot=True, fmt='d', cmap='YlGnBu')
plt.title('12. Heat Map: Passengers by Month and Year')
plt.show()
```

### 14. Correlation Map
Insight: Examines linear relationships between variables in diamonds data.


```python
plt.figure(figsize=(8, 6))
numeric_cols = diamonds.select_dtypes(include=[np.number])
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('14. Correlation Map: Diamonds Numeric Features')
plt.show()
```

### 16. Scatter Plot
Insight: Visual correlation between two continuous variables (carat vs price).


```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=diamonds.sample(1000, random_state=42), x='carat', y='price', alpha=0.5)
plt.title('16. Scatter Plot: Carat vs Price')
plt.show()
```

### 18. Bubble Chart
Insight: Scatter plot where dot size represents a 3rd variable.


```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x='total_bill', y='tip', size='size', sizes=(20, 200), hue='sex', alpha=0.7)
plt.title('18. Bubble Chart: Bill vs Tip (Size = Party Size)')
plt.show()
```

### 19. Lollipop Chart
Insight: Less cluttered alternative to the bar plot.


```python
tip_day = tips.groupby('day', observed=False)['tip'].mean().reset_index()
plt.figure(figsize=(8, 5))
plt.stem(tip_day['day'], tip_day['tip'], basefmt=" ")
plt.title('19. Lollipop Chart: Average Tip by Day')
plt.ylabel('Average Tip')
plt.ylim(0, 4)
plt.show()
```

### 20. Radar Chart
Insight: Compares variables of different magnitudes symmetrically.


```python
categories = ['carat', 'depth', 'table', 'x', 'y', 'z']
N = len(categories)
values = diamonds[categories].mean().values.tolist()
values += values[:1]
angles = [n / float(N) * 2 * np.pi for n in range(N)]
angles += angles[:1]

plt.figure(figsize=(6, 6))
ax = plt.subplot(111, polar=True)
plt.xticks(angles[:-1], categories)
ax.plot(angles, values, linewidth=1, linestyle='solid')
ax.fill(angles, values, 'b', alpha=0.1)
plt.title('20. Radar Chart: Average Attributes of Diamonds')
plt.show()
```

### 21. Violin Plot
Insight: Combines KDE and box plot metrics into one.


```python
plt.figure(figsize=(10, 6))
sns.violinplot(data=diamonds.sample(5000, random_state=42), x='cut', y='price', palette='Set3')
plt.title('21. Violin Plot: Price Distribution by Cut')
plt.show()
```

### 22. Box Plot
Insight: Five-number summary and outliers clearly displayed.


```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x='day', y='total_bill', palette='pastel')
plt.title('22. Box Plot: Total Bill by Day')
plt.show()
```

### 24. Waterfall Chart
Insight: Shows cumulative impacts of positive/negative values.


```python
changes = [100, 50, -30, -20, 60]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
cumulative = np.cumsum([0] + changes[:-1])
plt.figure(figsize=(8, 5))
colors = ['green' if c > 0 else 'red' for c in changes]
plt.bar(months, changes, bottom=cumulative, color=colors)
plt.title('24. Waterfall Chart: Profit Changes')
plt.ylabel('Profit')
plt.show()
```

### 25. Ridge Plot
Insight: Highly aesthetic way to compare multiple distributions.


```python
g = sns.FacetGrid(diamonds.sample(5000, random_state=42), row='cut', aspect=4, height=1.5)
g.map(sns.kdeplot, 'depth', fill=True, alpha=0.7, lw=1.5)
g.map(sns.kdeplot, 'depth', color='k', lw=1.5)
g.set_titles(row_template='{row_name}')
plt.subplots_adjust(hspace=-0.4)
plt.suptitle('25. Ridge Plot: Diamond Depth by Cut', y=1.02)
plt.show()
```

### 28. Swarm Plot
Insight: Non-overlapping categorical scatter plot.


```python
plt.figure(figsize=(8, 5))
sns.swarmplot(data=tips, x='day', y='tip', size=4, palette='deep')
plt.title('28. Swarm Plot: Tips by Day')
plt.show()
```

### 31. Area Chart
Insight: Emphasizes volume over time compared to a simple line.


```python
plt.figure(figsize=(10, 5))
plt.fill_between(flights_sum['year'], flights_sum['passengers'], color='skyblue', alpha=0.6)
plt.plot(flights_sum['year'], flights_sum['passengers'], color='blue', alpha=0.8)
plt.title('31. Area Chart: Total Passengers by Year')
plt.xlabel('Year')
plt.ylabel('Passengers')
plt.show()
```

### 32. Stacked Area Chart
Insight: Volume contribution of sub-categories over time.


```python
x = flights['year'].unique()
y1 = flights[flights['month']=='Jan']['passengers'].values
y2 = flights[flights['month']=='Feb']['passengers'].values
y3 = flights[flights['month']=='Mar']['passengers'].values
plt.figure(figsize=(8, 5))
plt.stackplot(x, y1, y2, y3, labels=['Jan', 'Feb', 'Mar'], colors=['#ff9999','#66b3ff','#99ff99'])
plt.legend(loc='upper left')
plt.title('32. Stacked Area Chart: Passengers for Q1 Months')
plt.show()
```

### 41. Hexbin Map
Insight: Excellent for continuous 2D densities with large datasets.


```python
plt.figure(figsize=(8, 6))
plt.hexbin(diamonds['table'], diamonds['depth'], gridsize=30, cmap='Blues')
plt.colorbar(label='Count')
plt.title('41. Hexbin Map: Table vs Depth')
plt.xlabel('Table')
plt.ylabel('Depth')
plt.show()
```

### 49. Combo Chart
Insight: Combines line and bar on identical x-axis.


```python
fig, ax1 = plt.subplots(figsize=(8, 5))
ax2 = ax1.twinx()
sns.barplot(data=tip_day, x='day', y='tip', ax=ax1, color='lightblue', alpha=0.7)
ax2.plot(range(len(tip_day)), tip_day['tip'] * 5, color='red', marker='o', label='Total Bill (Scaled)')
ax1.set_ylabel('Average Tip', color='blue')
ax2.set_ylabel('Scaled Total Bill', color='red')
plt.title('49. Combo Chart: Tips (Bar) and Scaled Bills (Line)')
plt.show()
```

### 50. Pareto Chart
Insight: Identifies the biggest factors contributing to the whole.


```python
counts = diamonds['cut'].value_counts().sort_values(ascending=False)
cum_perc = counts.cumsum() / counts.sum() * 100
fig, ax1 = plt.subplots(figsize=(8, 5))
ax1.bar(counts.index, counts.values, color='C0')
ax2 = ax1.twinx()
ax2.plot(counts.index, cum_perc, color='C1', marker='D', ms=7)
ax1.set_ylabel('Count', color='C0')
ax2.set_ylabel('Cumulative Percentage (%)', color='C1')
plt.title('50. Pareto Chart: Diamond Cuts')
plt.show()
```

### 53. Slope Chart
Insight: Highlights relative change between strictly two points in time.


```python
years = [1950, 1960]
val1 = [flights[flights['year']==1950]['passengers'].sum(), flights[flights['year']==1960]['passengers'].sum()]
plt.figure(figsize=(4, 6))
plt.plot(years, val1, marker='o', linewidth=2)
plt.xticks(years)
plt.title('53. Slope Chart: Passenger Growth 1950 vs 1960')
plt.ylabel('Total Passengers')
plt.show()
```

### 55. Column Chart
Insight: Simply another name for a vertical bar plot.


```python
plt.figure(figsize=(8, 5))
tips.groupby('day', observed=False)['tip'].sum().plot(kind='bar', color='coral')
plt.title('55. Column Chart: Total Tips by Day')
plt.ylabel('Total Tip Amount')
plt.xticks(rotation=0)
plt.show()
```

### 59. Q-Q Plot
Insight: Evaluates visually if data follows a normal distribution.


```python
plt.figure(figsize=(6, 6))
stats.probplot(diamonds['carat'].sample(1000, random_state=42), dist='norm', plot=plt)
plt.title('59. Q-Q Plot: Diamond Carats vs Normal Distribution')
plt.show()
```

### 61. Strip Plot
Insight: 1-dimensional scatter plot, highlights point densities without jitter.


```python
plt.figure(figsize=(8, 5))
sns.stripplot(data=tips, x='day', y='total_bill', palette='Set1', jitter=False)
plt.title('61. Strip Plot (No Jitter): Total Bill by Day')
plt.show()
```

### 62. Jitter Plot
Insight: Adds random noise to strip plot for visibility.


```python
plt.figure(figsize=(8, 5))
sns.stripplot(data=tips, x='day', y='total_bill', palette='Set1', jitter=True, alpha=0.7)
plt.title('62. Jitter Plot: Total Bill by Day')
plt.show()
```
